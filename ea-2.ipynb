```python
import numpy as np
import matplotlib.pyplot as plt
import csv


from RectanglesEA import CuttingStockEA, RectangleDef
```

```python
def wczytaj_dane(nazwa_pliku):
    try:
        promien = float(nazwa_pliku.split('.')[0][1:])
    except ValueError:
        print("Nie można odczytać promienia z nazwy pliku.")
        return None, []
        
    rect_types = []
    with open("data/"+nazwa_pliku, mode='r') as file:
        reader = csv.reader(file)
        for i, row in enumerate(reader):
            # Format: szerokość, wysokość, wartość
            w, h, v = float(row[0]), float(row[1]), float(row[2])
            rect_types.append(RectangleDef(i, w, h, v))
            
    return promien, rect_types
```

```python
r, typy_prostokatow = wczytaj_dane("r800.csv")
```

```python
ea = CuttingStockEA(radius=r, rect_types=typy_prostokatow, population_size=100)
```

```python
def zadanie(plik, epochs):
    r, typy_prostokatow = wczytaj_dane(plik)
    ea = CuttingStockEA(radius=r, rect_types=typy_prostokatow, population_size=100)
    print("Max chromosome len: ", ea.chrom_length)
    najlepszy_wynik, ulozenie = ea.run(generations=epochs)
    print("Koniec! Najlepsza znaleziona wartość:", najlepszy_wynik)
    ea.plot_solution(ulozenie)
```
